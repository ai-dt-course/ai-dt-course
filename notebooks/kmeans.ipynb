# K-Means Clustering from Scratch

This notebook implements K-Means clustering algorithm from scratch.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
```

## 1. Generate Sample Data

```python
# Generate sample data
X, y_true = make_blobs(n_samples=300, centers=3, cluster_std=0.8, random_state=42)

plt.scatter(X[:, 0], X[:, 1], c=y_true, cmap='viridis')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.title('Sample Data for Clustering')
plt.show()
```

## 2. Implement K-Means

```python
def initialize_centroids(X, k):
    """Initialize centroids randomly"""
    indices = np.random.choice(len(X), k, replace=False)
    return X[indices]

def compute_distances(X, centroids):
    """Compute Euclidean distance between each point and centroids"""
    distances = np.zeros((len(X), len(centroids)))
    for i, centroid in enumerate(centroids):
        distances[:, i] = np.sqrt(np.sum((X - centroid) ** 2, axis=1))
    return distances

def assign_clusters(X, centroids):
    """Assign each point to nearest centroid"""
    distances = compute_distances(X, centroids)
    return np.argmin(distances, axis=1)

def update_centroids(X, labels, k):
    """Update centroids based on cluster means"""
    centroids = np.zeros((k, X.shape[1]))
    for i in range(k):
        cluster_points = X[labels == i]
        if len(cluster_points) > 0:
            centroids[i] = np.mean(cluster_points, axis=0)
    return centroids

def kmeans(X, k, max_iterations=100):
    """K-Means algorithm"""
    # Initialize centroids
    centroids = initialize_centroids(X, k)
    
    for i in range(max_iterations):
        # Assign clusters
        labels = assign_clusters(X, centroids)
        
        # Update centroids
        new_centroids = update_centroids(X, labels, k)
        
        # Check convergence
        if np.allclose(centroids, new_centroids):
            print(f"Converged after {i+1} iterations")
            break
        
        centroids = new_centroids
    
    return centroids, labels
```

## 3. Run K-Means

```python
# Run K-Means
k = 3
centroids, labels = kmeans(X, k, max_iterations=100)

# Plot results
plt.scatter(X[:, 0], X[:, 1], c=labels, cmap='viridis')
plt.scatter(centroids[:, 0], centroids[:, 1], c='red', marker='X', s=200, label='Centroids')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.title('K-Means Clustering Result')
plt.legend()
plt.show()
```
