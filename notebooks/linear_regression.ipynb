# Linear Regression from Scratch

This notebook implements linear regression from scratch using gradient descent.

```python
import numpy as np
import matplotlib.pyplot as plt
```

## 1. Generate Sample Data

```python
# Generate sample data
np.random.seed(42)
X = 2 * np.random.rand(100, 1)
y = 4 + 3 * X + np.random.randn(100, 1)

plt.scatter(X, y)
plt.xlabel('X')
plt.ylabel('y')
plt.title('Sample Data')
plt.show()
```

## 2. Implement Linear Regression

```python
def compute_loss(X, y, w, b):
    """Compute Mean Squared Error"""
    m = len(y)
    predictions = X * w + b
    loss = (1/m) * np.sum((predictions - y) ** 2)
    return loss

def gradient_descent(X, y, w, b, learning_rate, iterations):
    """Perform gradient descent"""
    m = len(y)
    loss_history = []
    
    for i in range(iterations):
        predictions = X * w + b
        
        # Compute gradients
        dw = (2/m) * np.sum(X * (predictions - y))
        db = (2/m) * np.sum(predictions - y)
        
        # Update parameters
        w = w - learning_rate * dw
        b = b - learning_rate * db
        
        # Record loss
        loss = compute_loss(X, y, w, b)
        loss_history.append(loss)
        
        if i % 100 == 0:
            print(f"Iteration {i}: Loss = {loss:.4f}")
    
    return w, b, loss_history
```

## 3. Train the Model

```python
# Initialize parameters
w = 0
b = 0
learning_rate = 0.01
iterations = 1000

# Train
w_final, b_final, loss_history = gradient_descent(X, y, w, b, learning_rate, iterations)

print(f"\nFinal weight: {w_final[0][0]:.4f}")
print(f"Final bias: {b_final[0]:.4f}")
```

## 4. Visualize Results

```python
# Plot the fit line
plt.scatter(X, y, label='Data')
X_line = np.linspace(0, 2, 100).reshape(100, 1)
y_line = X_line * w_final + b_final
plt.plot(X_line, y_line, color='red', label='Fit Line')
plt.xlabel('X')
plt.ylabel('y')
plt.title('Linear Regression Result')
plt.legend()
plt.show()
```

## 5. Plot Loss History

```python
plt.plot(loss_history)
plt.xlabel('Iteration')
plt.ylabel('Loss (MSE)')
plt.title('Loss over Iterations')
plt.show()
```
